```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Generate data
test_data = np.linspace(0, 40, 1000)
d_s = 3 * (test_data - 5)
d_r = 2.5 * test_data

# Create the figure and axis
fig, axis = plt.subplots()
plt.title('A Bank Robber Caught')
plt.xlabel('Time in Minutes')
plt.ylabel('Distance in Kilometres')

# Plot both lines
axis.plot(test_data, d_r, c='green', label='Robber')
axis.plot(test_data, d_s, c='red', label='Sheriff')

plt.axvline(x=30, color='purple', linestyle = '--')
plt.axhline(y=75, color='purple', linestyle = '--')

# Add labels, legend, and show
axis.legend()
plt.show()
```
